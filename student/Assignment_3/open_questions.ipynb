The MLP is fundamentally limited by its reliance on mean-pooled FastText embeddings. Mean pooling reduces an entire sentence to a single averaged vector, which means the model effectively sees the sentence as a “bag of words” rather than a structured sequence. Information about word order and how words interact is lost in this process. As a result, the MLP struggles with sentiment that depends on context. This loss of structure makes it harder for the MLP to distinguish subtle sentiment differences, leading to weaker generalization and higher misclassification rates.

The LSTM addresses these limitations by processing text sequentially rather than all at once. Instead of averaging words, the LSTM reads the sentence word by word and maintains a memory of what it has seen so far. This allows it to track how sentiment develops across the sentence and to understand how earlier words influence the meaning of later ones. Because of this, the LSTM can better capture phrases where sentiment depends on word order or context, which helps explain its stronger validation and test performance.

Fine-tuned large language models such as BERT and GPT outperform both the MLP and classical sequence models because they start with a much higher understanding of language. These models are pretrained on massive and diverse text corpora, allowing them to learn nuanced patterns and context. When fine-tuned, they can apply this prior knowledge to recognize subtle sentiment cues, handle long-range dependencies, and figure out ambiguous phrasing far more effectively than models trained from scratch or with static embeddings.

From my test results, the models ranked in performance as GPT, BERT, LSTM, GRU, MLP, and RNN. Though, this ranking may vary slightly across runs due to randomness in initialization and training. Both GPT and BERT performed well due to being transformer-based models (having a unidirectional and bidirectional attention mechanism respectively), and their test performances are similar. As for the classical baselines, the MLP model should've performed the worst as it does not model sequences at all, but RNN suffers from vanishing or exploding gradients which can lead to less stable predictions resulting in a lower test score (even if RNN explicity models sequential order). GRU improves upon the RNN and MLP by introducing gating mechanisms that control how information flows through time. These gates allow the model to retain relevant information longer and forget irrelevant signals. LSTM performs better than the three other classical models as further it extends upon the idea of GRU with a more expressive memory cell and separate input, forget, and output gates. This allows it to model longer-range dependencies and more complex patterns.
